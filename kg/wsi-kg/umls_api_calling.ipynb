```python
import requests

# UMLS API configuration
API_KEY = "21b2a93f-3178-4807-b2e1-513bdbf1af8e"  # Replace with your UMLS API key
BASE_URL = "https://uts-ws.nlm.nih.gov/rest"
```

```python
def get_umls_cui(search_term, api_key=API_KEY):
    """
    Search for a term in UMLS and return matching CUIs
    
    Args:
        search_term: The medical term to search for
        api_key: Your UMLS API key
    
    Returns:
        List of dictionaries containing CUI and term information
    """
    search_url = f"{BASE_URL}/search/current"
    
    params = {
        "string": search_term,
        "apiKey": api_key,
        "returnIdType": "concept"
    }
    
    try:
        response = requests.get(search_url, params=params)
        response.raise_for_status()
        
        data = response.json()
        results = []
        
        if "result" in data and "results" in data["result"]:
            for item in data["result"]["results"]:
                results.append({
                    "cui": item.get("ui"),
                    "name": item.get("name"),
                    "source": item.get("rootSource")
                })
        
        return results
    
    except requests.exceptions.RequestException as e:
        print(f"Error calling UMLS API: {e}")
        return []

# Example usage
search_term = "eosinophilic cytoplasm"
cuis = get_umls_cui(search_term)

for result in cuis:
    print(f"CUI: {result['cui']}, Name: {result['name']}, Source: {result['source']}")
```

```python
def get_cui_definitions(cui, api_key=API_KEY):
    url = f"{BASE_URL}/content/current/CUI/{cui}/definitions?language=ENG"

    params = {
        "apiKey": api_key
    }

    try:
        response = requests.get(url, params=params)
        response.raise_for_status()

        data = response.json()
        definitions = []

        if "result" in data:
            for item in data["result"]:
                    definitions.append({
                        "definition": item.get("value"),
                        "source": item.get("rootSource")
                    })

        return definitions

    except requests.exceptions.RequestException as e:
        print(f"Error fetching definitions for {cui}: {e}")
        return []
    


search_term = "renal cell carcinoma"

results = get_umls_cui(search_term)

for res in results:
    cui = res["cui"]
    name = res["name"]

    print("\n====================")
    print(f"Term: {search_term}")
    print(f"CUI: {cui}")
    print(f"Name: {name}")
    print(f"Source: {res['source']}")

    definitions = get_cui_definitions(cui)

    if definitions:
        for d in definitions:
            print(f"Definition ({d['source']}): {d['definition']}")
    else:
        print("No definition found")
```

```python
import requests
import time

def get_atoms_snomed(cui, api_key=API_KEY):
    url = f"{BASE_URL}/content/current/source/SNOMEDCT_US/{cui}/atoms?language=ENG"

    params = {
        "apiKey": api_key
    }

    try:
        response = requests.get(url, params=params)
        response.raise_for_status()

        data = response.json()
        atoms = []

        if "result" in data:
            for item in data["result"]:
                atoms.append({
                    "aui": item.get("ui"),         
                    "name": item.get("name"),
                    "termType": item.get("termType"), 
                    "code": item.get("code"),         
                    "source": item.get("rootSource")
                })

        return atoms

    except requests.exceptions.RequestException as e:
        print(f"Error fetching atoms for {cui}: {e}")
        return []
    


search_term = "proliferative index"

results = get_umls_cui(search_term)

for res in results:
    cui = res["cui"]
    name = res["name"]

    print("\n====================")
    print(f"Term: {search_term}")
    print(f"CUI: {cui}")
    print(f"Name: {name}")
    print(f"Source: {res['source']}")

    definitions = get_cui_definitions(cui)

    if definitions:
        for d in definitions:
            print(f"Definition ({d['source']}): {d['definition']}")
    else:
        print("No definition found")


    atoms = get_atoms_snomed(cui)
    print(atoms)
    if atoms:
        print("Atoms:")
        for a in atoms:
            print(f" - {a['name']} (Source: {a['source']})")
    else:
        print("No atoms found")

```
